````python
# Solución — Bloque 1 (Conjuntos)

```python
# P1: newsletter → unión
newsletter = oyentes_pop | oyentes_rock
print("Newsletter a:", sorted(newsletter), "→", len(newsletter), "usuarios")

# P2: fidelización → intersección y diferencia
ambos = oyentes_pop & oyentes_rock
print("Escuchan ambos:", sorted(ambos))
solo_pop = oyentes_pop - oyentes_rock
print("Pop sin rock (oferta rock):", sorted(solo_pop))

# P3: recomendaciones → producto cartesiano
from itertools import product
parejas = list(product(usuarios_todos, {"pop", "rock"}))
print(len(parejas), "parejas;", "oportunidad:", ("Pedro", "pop"))
```

# Solución — Bloque 2 (Relaciones)

```python
# P4: simetría y pares rotos
def es_simetrica(rel):
    return all((b, a) in rel for (a, b) in rel)

def es_reflexiva(rel, conjunto):
    return all((x, x) in rel for x in conjunto)

print("¿Simétrica?", es_simetrica(sigue))
rotos = [(a, b) for (a, b) in sigue if (b, a) not in sigue]
print("Pares rotos:", rotos)
print("No correspondidos:", sorted({a for a, b in rotos}))
print("¿Reflexiva?", es_reflexiva(sigue, usuarios_todos))  # False: nadie se sigue a sí mismo

# P5: lista de adyacencia
adyacencia = {u: set() for u in usuarios_todos}
for a, b in sigue:
    adyacencia[a].add(b)
print(adyacencia)
print("Elena sigue a:", adyacencia["Elena"])
seguidos = {}
for a, b in sigue:
    seguidos[b] = seguidos.get(b, 0) + 1
print("Más seguida:", max(seguidos, key=seguidos.get), seguidos)
```

# Solución — Bloque 3 (Funciones)

```python
# P6
canciones = {u["nombre"]: u["canciones"] for u in usuarios}
def f(u):  return canciones.get(u, None)      # función parcial
def g(c):  return 100 - c // 5

engagement = {u: g(f(u)) for u in canciones}
print(engagement)
print(f("César: {f('César')}")  # None → función parcial (fuera del dominio)
```

# Solución — Bloque 4 (Lógica)

```python
# P7
premium = [u["nombre"] for u in usuarios
           if u["edad"] >= 25 and u["canciones"] > 100 and u["plan"] == "gratis"]
retencion = [u["nombre"] for u in usuarios
             if u["plan"] == "gratis" and not (u["canciones"] > 50)]
print(premium, retencion)
# (c): >100 y NOT >50 no pueden cumplirse a la vez → listas disjuntas
```

# Solución — Bloque 5 (Grafos)

```python
# P8
import networkx as nx, matplotlib.pyplot as plt
G = nx.DiGraph()
G.add_nodes_from(usuarios_todos)
G.add_edges_from(sigue)

in_deg = dict(G.in_degree())
print("Seguidores:", in_deg)
print("Influencer:", max(in_deg, key=in_deg.get))       # Ana, 3 seguidores

pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_size=800, node_color="#ffd89f",
        arrows=True, arrowsize=20)
plt.show()
# (d): se pierden las 3 flechas que apuntan a Ana
```

# Solución — Bloque 6 (Combinatoria)

```python
# P9
from itertools import combinations, permutations
artistas = ["Dua Lipa", "Rosalía", "Bad Bunny", "Quevedo", "Karol G"]
print(len(list(combinations(artistas, 2))))    # 10 = 5!/(2!·3!)
print(len(list(permutations(artistas, 2))))    # 20 = 5·4
# (c): cada pareja {A,B} genera 2 órdenes (A antes de B, B antes de A)
```

# Solución — Bloque 7 (Aritmética modular)

```python
# P10
letras = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
def cifrar(m, k=5):
    return "".join(letras[(letras.index(c) + k) % 26] for c in m.upper())
def descifrar(m, k=5):
    return "".join(letras[(letras.index(c) - k) % 26] for c in m.upper())

print(cifrar("BONUS"))    # GTSZX
print(descifrar("MJQQT")) # HELLO
print(cifrar("ZAPATO"))   # EFUFYT → la Z (25+5=30) vuelve a ser E gracias a % 26
```

# Solución — Bloque 8 (pandas)

```python
import pandas as pd
df = pd.DataFrame(usuarios)
print(df.groupby("plan")["canciones"].sum())               # gratis 315, premium 1030
print(df[df["nombre"].isin(oyentes_pop)]["edad"].mean())   # 29.0
# (c): un DataFrame es una RELACIÓN (conjunto de tuplas con atributos)
```
````
