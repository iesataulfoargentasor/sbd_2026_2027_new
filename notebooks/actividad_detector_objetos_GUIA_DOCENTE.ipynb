## Respuestas esperadas

**Pregunta 1(c):** la dimensión floja suele ser *fondos* o *iluminación*. Consecuencia esperada: el detector solo funcionará en el entorno donde se fotografió (p. ej. solo sobre la mesa del aula). Se pide conectar dimensión → fallo previsible.

**Pregunta 2:**
- (a) Una época = el modelo ve TODAS las imágenes de entrenamiento una vez. 40 épocas = 40 pasadas.
- (b) YOLOv8n es el modelo Nano preentrenado sobre millones de imágenes (COCO). Empezar de él es **transfer learning**: reutiliza características visuales ya aprendidas (bordes, texturas, formas) y solo ajusta la capa final a nuestra clase. Entrenar de cero con 100 fotos es imposible.
- (c) `valid/` sirve para medir el rendimiento en datos que no se usan para ajustar los pesos. Si train es altísimo y valid bajo → **sobreajuste (overfitting)**: el modelo memorizó las fotos en vez de generalizar.
- (d) mAP50 típico con 60-100 fotos buenas: 0.70-0.95. Por debajo de 0.5, revisar etiquetas/dataset.

**Pregunta 3:**
- (a) No tiene por qué coincidir: mAP50 se mide sobre el conjunto de validación (que Roboflow sacó de las MISMAS fotos subidas); las 10 manuales son un test más exigente y pequeño (poca muestra → porcentaje variable).
- (b) Cualquier fallo documentado vale si lo conectan con la dimensión floja (p. ej. no detecta a distancia → pocas fotos lejanas).
- (c) **Falso positivo** por confusión con un objeto visualmente parecido. Solución: añadir al dataset fotos del objeto parecido etiquetadas como fondo/negativo o como otra clase (esto es *hard negative mining* simplificado).

**Pregunta 4:**
- (a) Se espera un ejemplo del estilo: etiqueta mal dibujada / caja que incluía el fondo → el modelo aprendió mal ese caso.
- (b) Cualquier dos de: ingestión masiva (no subir a mano: Flume/Sqoop/Kafka), almacenamiento distribuido (HDFS/cluster, no un disco), procesamiento en paralelo (no un solo PC: Spark), coste de GPU en la nube, monitorización del sistema. La clave es que identifiquen el salto PC → sistema distribuido.
- (c) Cualquier dos de: RGPD/datos de salud especiales (consentimiento, anonimización), el sistema asiste pero no sustituye al profesional (responsabilidad del diagnóstico), sesgos del dataset (ecógrafos distintos, poblaciones distintas), validar con datos del propio hospital, transparencia/explicabilidad ante errores.

## Consejos de gestión de aula

- **Roboflow**: plan gratuito "Public" (los datasets son visibles → avisad de no subir nada personal). Alternativa sin cuentas web: LabelImg en los equipos + organizar carpetas YOLO a mano (más laborioso pero didáctico).
- **Colab GPU**: límite de uso gratuito; si se agota, `yolov8n` con 40 épocas y ~150 imágenes tarda ~10-15 min en GPU T4, así que escalated por turnos si hace falta. CPU pura: varias horas (como el artículo: 9,5 h) — buen momento para recordar Big-O.
- **El error nº 1** de los grupos es hacer 60 fotos idénticas del objeto en la mesa. La checklist de la Fase 1 es la prevención; el fallo en la Fase 4 es el aprendizaje.
- **Presentación final:** pedid que muestren UN fallo además de los aciertos; evalúa mejor la reflexión que la demo.
